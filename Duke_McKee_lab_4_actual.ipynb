```python
import matplotlib.pyplot as plt
import numpy as np

# DATA TO BE PLOTTED
beta_kg = np.array(
    [0.00381472, 0.00803444, 0.00699423, 0.01196890, 0.01087521, 0.01252866]
)  # units: kg

hanging_mass_mh_kg = np.array(
    [0.005, 0.006, 0.007, 0.008, 0.009, 0.010]
)  # units: kg

err_hanging_mass_mh_kg = np.array(
    [0.0001, 0.0001, 0.0001, 0.0001, 0.0001, 0.0001]
)  # units: kg

# Re-assign variables as x, y, dy
x = beta_kg
y = hanging_mass_mh_kg
dy = err_hanging_mass_mh_kg

# Plot Labels
plot_title = "Work and Kinetic Energy: Hanging Mass vs. Beta"
x_label = "Beta [kg]"
y_label = "Hanging Mass m_h [kg]"

# ----------------------------------------------------------------------- #
# ----- (LINEAR FIT) ----- #

# Size the plot
plt.figure(figsize=(12, 8))

# Create plot with uncertainty error bars
plt.errorbar(
    x,
    y,
    yerr=dy,
    fmt="o",
    color="blue",
    ecolor="red",
    capsize=3,
    label="Data Points",
)

# Labels and Title
plt.xlabel(x_label)
plt.ylabel(y_label)
plt.title(plot_title)
plt.grid(True)

# Fitting to a 1st degree polynomial (Linear: y = mx + b)
# np.polynomial.polyfit returns coefficients in increasing degree [intercept, slope]
c, b = np.polynomial.polynomial.polyfit(x, y, 1)

# Annotate with values of Slope (b) and Intercept (c)
plt.annotate("best fit line: y = mx + b", (0.05, 0.92), xycoords="axes fraction")
plt.annotate(
    "     m (slope) = {:.6f}".format(b), (0.05, 0.86), xycoords="axes fraction"
)
plt.annotate(
    "     b (intercept / m_f) = {:.6f} kg".format(c),
    (0.05, 0.80),
    xycoords="axes fraction",
)

# Create linear fit line
xnew = np.linspace(x.min(), x.max(), 300)
fit = b * xnew + c

plt.plot(xnew, fit, color="green", label="Linear Fit")
plt.legend()
plt.show()
```
